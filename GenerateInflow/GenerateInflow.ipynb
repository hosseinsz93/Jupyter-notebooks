The following script is to generate InflowPlane.dat file for the older version of the GenerateInflow.c code.

```python
# Generate a 2D inflow plane in Tecplot ASCII format
Nx_LES = 113   # Points in x-direction
Ny_LES = 113   # Points in y-direction
Lx = 1.0       # Domain length in x-spanwise
Ly = 1.0       # Domain length in y-vertical

with open("InflowPlane.dat", "w") as f:
    # Write header lines
    for i in range(10):
        f.write(f"# Header line {i+1}\n")
    
    # Write grid data (2D plane with Z=0)
    for j in range(Ny_LES):
        for i in range(Nx_LES):
            x = (i / (Nx_LES - 1)) * Lx
            y = (j / (Ny_LES - 1)) * Ly
            z = 0.0
            nvert = 0.0  # Example value (1=inside domain)
            f.write(f"{x} {y} {z} {nvert}\n")
```

The following script is to generate xyz file for the new code's inflow.

```python
Nx = 102  # Total grid points in x
Ny = 102  # Total grid points in y
Lx = 0.61  # Domain length in x
Ly = 0.61  # Domain length in y

with open("xyz.dat", "w") as f:
    # First line
    f.write(f"{Nx-1} {Ny-1} 0\n")

    # X-coordinates (Nx-1 values)
    for i in range(Nx-1):
        x = i * Lx / (Nx-2)  
        f.write(f"{x} 0.0 0.0\n")

    # Y-coordinates (Ny-1 values)
    for j in range(Ny-1):
        y = j * Ly / (Ny-2)  
        f.write(f"0.0 {y} 0.0\n")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import os

def plot_energy_spectrum():
    data = np.loadtxt('E_k.dat')
    k = data[:, 0]  # Wavenumber
    E = data[:, 1]  # Energy spectrum
    title = 'Energy Spectrum (from E_k.dat file)'
    
    # Create the plot (using log-log scale)
    plt.figure(figsize=(10, 6))
    plt.loglog(k, E, 'b-', linewidth=2, label='Energy Spectrum')
    plt.xlabel('Wavenumber (k)', fontsize=14)
    plt.ylabel('Energy Spectrum E(k)', fontsize=14)
    plt.title(title, fontsize=16)
    plt.grid(True, which="both", ls="-", alpha=0.2)
    
    # Add reference -5/3 slope (Kolmogorov spectrum in inertial range)
    k_ref = np.array([min(k[k>0])*5, max(k)/5])
    
    # Scale reference line to match energy spectrum at some point
    idx = len(k) // 3
    scale_factor = E[idx] / (k[idx]**(-5/3))
    ref_line = scale_factor * k_ref**(-5/3)
    
    plt.loglog(k_ref, ref_line, 'r--', linewidth=2, label='-5/3 Slope (Kolmogorov)')
    
    # Add annotation showing the formula
    formula = r'$E(k) = \alpha \epsilon^{2/3} L^{5/3} \frac{L^4 k^4}{(1+(Lk)^2)^{17/6}}$'
    plt.annotate(formula, xy=(0.5, 0.05), xycoords='axes fraction', 
                bbox=dict(boxstyle="round,pad=0.5", fc="white", alpha=0.8),
                ha='center', fontsize=12)
    
    plt.legend()
    plt.tight_layout()
    plt.savefig('energy_spectrum.png', dpi=300)
    plt.show()

if __name__ == "__main__":
    plot_energy_spectrum()
```
